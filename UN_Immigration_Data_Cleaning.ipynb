```python
import pandas as pd
from IPython.display import display

print("pandas version:", pd.__version__)

try:
    from google.colab import files
    uploaded = files.upload()
except ImportError:
    print("Local Jupyter detected. Make sure Canada.xlsx is in the current folder.")

file_path = "Canada.xlsx"
print("Input file:", file_path)

df = pd.read_excel(
    file_path,
    sheet_name="Canada by Citizenship",
    header=20
)

print("Original table shape:", df.shape)
display(df.head())
```

```python
# Use easier-to-read names for the country, continent, and region.
df = df.rename(columns={
    "OdName": "Country",
    "AreaName": "Continent",
    "RegName": "Region"
})

# Remove technical columns that are not part of the desired display.
columns_to_drop = ["Type", "Coverage", "AREA", "REG", "DEV"]
df = df.drop(columns=[c for c in columns_to_drop if c in df.columns])

# Identify the year columns and sort them from earliest to latest.
year_columns = sorted(
    [c for c in df.columns if str(c).isdigit()],
    key=int
)

# Add a Total column by summing each country's annual values.
df["Total"] = df[year_columns].sum(axis=1)

# Put country details first, then annual values, then Total.
details = ["Country", "Continent", "Region", "DevName"]
df = df[[c for c in details if c in df.columns] + year_columns + ["Total"]]

# Remove accidental leading/trailing spaces from text fields.
for col in ["Country", "Continent", "Region", "DevName"]:
    if col in df.columns:
        df[col] = df[col].astype("string").str.strip()

display(df.head())
```
